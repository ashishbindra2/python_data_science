```python
from cv2 import imread
from cv2 import waitKey
from cv2 import destroyAllWindows
from cv2 import imshow
```

```python
inv_img = imread('./invoices/page_11.png')
```

```python
from pytesseract import image_to_string
from pytesseract import image_to_data
```

```python

text_cv = image_to_string(inv_img)
text_cv
```

```python
data = image_to_data(inv_img)
data
```

```python
data_list = list(map(lambda x : x.split('\t'),data.split('\n')))

for _ in data_list:
    print(_)
```

```python
from pandas import DataFrame

df = DataFrame(data_list[1:], columns=data_list[0])
df
```

```python
df = df.dropna()
```

```python
col_int = ['level', 'page_num', 'block_num','par_num', 'line_num', 'word_num', 'left', 'top', 'width','height','conf']

df[col_int] = df[col_int].astype(float, copy=True).astype(int)
```

```python
df[['level', 'left', 'top', 'width', 'height', 'conf']]
```

```python
for l,x,y,w,h,c in df[['level', 'left', 'top', 'width', 'height', 'conf']].values:
    print(l,x,y,w,h,c)
```

```python
image = inv_img.copy()
```

```python
from cv2 import rectangle
from cv2 import putText
from cv2 import FONT_HERSHEY_PLAIN
from cv2 import resize
```

```python

level ='word'

for l, x, y, w, h, c, txt in df[['level', 'left', 'top', 'width', 'height', 'conf','text']].values:
    print(l,x,y,w,h,c)
    
    if level == 'page':
        if l == 1:
            rectangle(img=image, pt1=(x,y),pt2=(x+w,y+h), color=(0,0,0),thickness=2)
        else:
            continue
    elif level == 'block':
        if l == 2:
            rectangle(img=image, pt1=(x,y),pt2=(x+w,y+h), color=(255,0,0),thickness=2)
        else:
             continue
    elif level == 'para':
        if l == 3:
            rectangle(img=image, pt1=(x,y),pt2=(x+w,y+h), color=(0,255,0),thickness=2)
        else:
             continue
    elif level == 'line':
        if l == 4:
            rectangle(img=image, pt1=(x,y),pt2=(x+w,y+h), color=(0,0,255),thickness=2)
        else:
             continue
    elif level == 'word':
        if l == 5:
            rectangle(img=image, pt1=(x,y),pt2=(x+w,y+h), color=(0,255,255),thickness=2)
            putText(image,txt,(x,y),FONT_HERSHEY_PLAIN,2,(255, 0, 0),2)
        else:
             continue
imgg = resize(image, (1000, 1100))
imshow("Bounding Box", imgg)
waitKey()
destroyAllWindows()

```
