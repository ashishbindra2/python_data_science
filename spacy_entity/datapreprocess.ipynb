### Deleting Characters:
#### How str.maketrans() Works:
str.maketrans() creates a dictionary-like mapping of characters or strings, which can then be passed to translate() to perform the actual substitution.

```python
# Delete characters 'a', 'b', 'c'
trans = str.maketrans('', '', 'abc')

text = "abcxyz"
result = text.translate(trans)
print(result)  # Output: xyz

text = "acxybz"
result = text.translate(trans)
print(result)  # Output: xyz

text = "axyz"
result = text.translate(trans)
print(result)  # Output: xyz
```

```python
# Mapping 'a' to '1', 'b' to '2', and 'c' to '3'
trans = str.maketrans('abc', '123')

text = "abcabc"
result = text.translate(trans)
print(result)  # Output: 123123

text = "abcxyz"
result = text.translate(trans)
print(result)  # Output: 123123

text = "qwe a"
result = text.translate(trans)
print(result)  # Output: 123123
```

```python
white_space = whitespace
punch = '!"#$%&\'()*+,:;<=>?[\\]^_`{|}~'

table_white_space = str.maketrans('','', white_space)
table_puncuation = str.maketrans('','',punch)
```

```python
def clean_text(txt):
    text = str(txt)
    text = text.lower()

    remove_whitespace = text.translate(table_white_space)
    remove_punctation = remove_whitespace.translate(table_puncuation)

    return str(remove_punctation)
```

#### for all image data

```python
group = data_clean.groupby(by='id')
cards = group.groups.keys()
cards
```

```python
all_cards_data = []

for card in cards:
    card_data = []
    content = '' # all inforamtion in entire text
    annotation = {'entities' : []} # all information of labels
    start = 0
    end = 0 
    
    groups_arr = group.get_group(card)[['text', 'tag']].values

    for text, label in  groups_arr:
        # print(text, label)
        text = str(text)
        str_len = len(text) + 1
    
        start = end
        end = start + str_len
        # print(start,end)
    
        if label != 'O':
            annot = (start, end - 1, label)
            annotation['entities'].append(annot)
        
        content = content + text + ' ' 
    
    card_data = (content, annotation)
    all_cards_data.append(card_data)
```
