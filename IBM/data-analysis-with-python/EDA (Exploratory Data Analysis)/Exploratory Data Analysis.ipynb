### value_Counts()
- Summarize the Categorical data is by using value_counts() 
- the varable that can divied into different categories or group and have descreate values


```python
import pandas as pd
import numpy as np

df  = pd.read_csv("../data/automobile.csv")
df["drive-wheels"]
```

```python
drive_wheels_counts = df["drive-wheels"].value_counts()
drive_wheels_counts
```

```python
drive_wheels_counts.rename({"drive-wheels":"value_counts"}, inplace= True)
drive_wheels_counts.index.name = "drive-wheels"

drive_wheels_counts
```

### Box Plots
- Mean => middle point of data
- Upper Quartile => 75th percentile
- Lower Quartile = >25th percentile
- inter quartile range => data between the upper and lower quartile
- lower Extreme => 1.5 * inter-quartile range(IQR) below 25th percentile(Lower Quartile)
- Upper Extreme => 1.5 * inter-quartile range above 75th percentile(upper quartile)
- Outlier / Single data point = >  occure outside the upper and lower extremes

whith box plot you can easily spoit outliers
- also se distribution and skewness of data



### 

```python
import seaborn as sns
sns.set_theme()

sns.boxplot(x = "drive-wheels", y="price", data=df)
```
