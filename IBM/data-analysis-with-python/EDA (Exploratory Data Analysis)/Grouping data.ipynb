### Q Is there any relationship between the different types of drive system and price of the vhicles?


### Group all the data by the different types of drive wheels and compare the results of these different drive wheels against each other.
- Use Panda `dataframe.groupby()` method
    - Can be applied to categoricak variables
    - Group data into categories
    - Group into single or multiple variables
- 

```python
import pandas as pd
import numpy as np

df  = pd.read_csv("../data/automobile.csv")
df["drive-wheels"]
```

```python
df["price"].replace("?",np.nan,inplace=True)
df["price"].eq(np.nan).any()
```

```python
mean = df["price"].mean()
# df["price"].replace(np.nan,mean,inplace=True)
```

```python
df["body-style"]
```

```python
df_test = df[["drive-wheels", "body-style", "price"]]
df_test
```

```python
# groupby() :

# df_test = df[["drive-wheels", "body-style", "price"]]
df_grp = df_test.groupby(["drive-wheels", "body-style"], as_index=False).mean()

df_grp
```

### A table of this form is not the easiest to read and also not very easy to visualize
- to make it easier to understand
    - We can transform this table to a pivot table by using the pivot method
    - `df_pivot = df_grp.pivot(index="drive-wheels", columns = "body-style")`
- 

```python
df_pivot = df_grp.pivot(index="drive-wheels", columns = "body-style")
df_pivot
```
